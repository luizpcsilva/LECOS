```python
# o código abaixo agrega o consumo ativo sequencial e calcula a média por método

df_max = pd.read_csv("ativo-Rmax-sequencial.csv")
df_min = pd.read_csv("ativo-Rmin-sequencial.csv")

df_media_sequencial_max = df_max.groupby('método')['Média'].mean()
display(df_media_sequencial_max)
df_media_sequencial_max.to_csv("media_sequencial_max.csv")

df_media_sequencial_min = df_min.groupby('método')['Média'].mean()
display(df_media_sequencial_min)
df_media_sequencial_min.to_csv("media_sequencial_min.csv")
```

```python
# o código abaixo agrega o consumo ativo paralelo e calcula a média por método

df_max = pd.read_csv("ativo-Rmax-paralelo.csv")
df_min = pd.read_csv("ativo-Rmin-paralelo.csv")

df_media_paralelo_max = df_max.groupby('par')['Média'].mean()
df_media_paralelo_max['par'] = df_media_paralelo_max['par'].apply(ast.literal_eval)
display(df_media_paralelo_max)
df_media_paralelo_max.to_csv("media_paralelo_max.csv")

df_media_paralelo_min = df_min.groupby('par')['Média'].mean()
df_media_paralelo_min['par'] = df_media_paralelo_min['par'].apply(ast.literal_eval)
display(df_media_paralelo_min)
df_media_paralelo_min.to_csv("media_paralelo_min.csv")
```
